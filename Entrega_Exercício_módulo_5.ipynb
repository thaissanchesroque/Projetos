# Entrega: Exercício módulo 5

```python
historico = []

while True:
    print("\n=== CALCULADORA INTELIGENTE ===")

    try:
        num1 = float(input("Digite o primeiro número: "))
        num2 = float(input("Digite o segundo número: "))
    except:
        print("Erro: digite apenas números.")
        continue

    print("\nEscolha a operação:")
    print("1 - Soma")
    print("2 - Subtração")
    print("3 - Multiplicação")
    print("4 - Divisão")

    opcao = input("Digite a opção (1/2/3/4): ")

    if opcao == "1":
        resultado = num1 + num2
        operacao = "Soma"

    elif opcao == "2":
        resultado = num1 - num2
        operacao = "Subtração"

    elif opcao == "3":
        resultado = num1 * num2
        operacao = "Multiplicação"

    elif opcao == "4":
        if num2 != 0:
            resultado = num1 / num2
            operacao = "Divisão"
        else:
            print("Erro: divisão por zero!")
            continue
    else:
        print("Opção inválida!")
        continue

    print("Resultado:", resultado)

    historico.append({
        "num1": num1,
        "num2": num2,
        "operacao": operacao,
        "resultado": resultado
    })

    continuar = input("\nDeseja fazer outra operação? (s/n): ")
    if continuar.lower() != "s":
        break

print("\n=== RESUMO DAS OPERAÇÕES ===")
for item in historico:
    print(item)
```
